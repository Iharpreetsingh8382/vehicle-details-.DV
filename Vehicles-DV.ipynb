```python
import numpy as np
import pandas as pd
```

# Step 1: Load the Dataset

```python
# load the dataset
datas = pd.read_csv('vehicles_dataset.csv')
datas
```

```python
# display the info
datas.info()

# show the first five rows
datas.head(5)
```

# Step 2: Explore the Data

```python
# Check for missing values
print(datas.isnull().sum())

# Summary statistics
print(datas.describe())
```

# Step 3: Clean the Data

```python
# Drop rows with missing critical values
datas.dropna(subset=['price', 'mileage'], inplace=True)

# Convert categorical columns if necessary
datas['type'] = datas['type'].astype(str)
```

```python
datas
```

# Step 4: Import Visualization Libraries

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

# Step 5: Import Visualization Libraries

```python
plt.figure(figsize=(10, 6))
sns.histplot(datas['price'], bins=50, kde=True)
plt.title("Distribution of Vehicle Prices")
plt.xlabel("Price")
plt.ylabel("Count")
plt.show()
```

# Step 6: Distribution of Mileage

```python
plt.figure(figsize=(10, 6))
sns.histplot(datas['mileage'], bins=50, kde=True)
plt.title("Distribution of Vehicle Mileage")
plt.xlabel("Mileage")
plt.ylabel("Count")
plt.show()
```

# Step 7: Boxplot of Price by Vehicle Type

```python
plt.figure(figsize=(12, 6))
sns.boxplot(data=datas, x='type', y='price')
plt.xticks(rotation=45)
plt.title("Price Distribution by Vehicle Type")
plt.show()
```

# Step 8: Scatter Plot (Mileage vs Price)

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(data=datas, x='mileage', y='price', hue='type', alpha=0.5)
plt.title("Price vs. Mileage across Vehicle Types")
plt.xlabel("Mileage")
plt.ylabel("Price")
plt.legend()
plt.show()
```

# Step 9: Pairplot for Important Features

```python
sns.pairplot(datas[['price', 'mileage', 'type']], hue='type')
plt.show()
```
